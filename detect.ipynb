```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
model.train(
 data="plastic_dataset/data.yaml",
    epochs=150,
    patience=25,   
    imgsz=640,
    batch=16,       
    optimizer="AdamW", 
    lr0=0.001, 
    cos_lr=True,
    weight_decay=0.001,        
    degrees=10, 
    translate=0.2, 
    scale=0.6, 
    fliplr=0.5, 
    flipud=0.2,
    mixup=0.1, 
    copy_paste=0.3,
    close_mosaic=15,
)
```
